# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

```python
# TODO
cols = ['age', 'fare', 'family_size']

print("Giá trị trung bình (Mean):")
print(df[cols].mean())

print("\nGiá trị trung vị (Median):")
print(df[cols].median())

print("\nGiá trị xuất hiện nhiều nhất :")
print(df[cols].mode())
```

## Group 2 — Dispersion

```python
# TODO
print("Độ lệch chuẩn (Standard Deviation):")
print(df[cols].std())

print("\nPhương sai (Variance):")
print(df[cols].var())

print("\nKhoảng biến thiên (Range = Max - Min):")
print(df[cols].max() - df[cols].min())

print("\nKhoảng tứ phân vị (IQR = Q3 - Q1):")
print(df[cols].quantile(0.75) - df[cols].quantile(0.25))
```

## Group 3 — Location and Shape

```python
# TODO
print("Độ lệch (Skewness):")
print(df[cols].skew())

print("\nĐộ nhọn (Kurtosis):")
print(df[cols].kurt())

print("\nCác mức phân vị (Q1=25%, Q2=50%, Q3=75%):")
print(df[cols].quantile([0.25, 0.5, 0.75]))
```

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

```python
# TODO
pclass_survival = df.groupby('pclass')['survived'].mean()
print("Tỷ lệ sống sót theo hạng vé:")
print(pclass_survival)

diff = pclass_survival.max() - pclass_survival.min()
print(f"\nHạng {pclass_survival.idxmax()} cao nhất ({pclass_survival.max():.2%}) ")
print(f"=> Chênh lệch: {diff * 100:.2f}%")
```

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

```python
# TODO
sex_survival = df.groupby('sex')['survived'].mean()
sex_diff = sex_survival.max() - sex_survival.min()

pclass_survival = df.groupby('pclass')['survived'].mean()
pclass_diff = pclass_survival.max() - pclass_survival.min()

print("Tỷ lệ sống sót theo giới tính:")
print(sex_survival)
print(f"-> Chênh lệch theo giới tính: {sex_diff * 100:.2f}%\n")

print("Tỷ lệ sống sót theo hạng vé:")
print(pclass_survival)
print(f"-> Chênh lệch theo hạng vé: {pclass_diff * 100:.2f}%\n")
```

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

```python
# TODO
print("Giá vé trung bình theo tình trạng sống sót:")
print(df.groupby('survived')['fare'].mean())

print(f"\nHệ số tương quan (fare & survived): {df['fare'].corr(df['survived']):.4f}")
```

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

```python
# TODO
family_survival = df.groupby('family_size')['survived'].agg(['count', 'mean'])
display(family_survival)
```

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

```python
# TODO
embark_survival = df.groupby('embark_town')['survived'].agg(['count', 'mean']).sort_values(by='mean', ascending=False)
display(embark_survival)
```
